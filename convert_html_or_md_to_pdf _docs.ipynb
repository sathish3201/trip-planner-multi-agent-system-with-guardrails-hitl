```python
from weasyprint import HTML

html_string = """
<html>
  <body>
    <h1>Hello</h1>
    <p>This is a PDF generated from HTML.</p>
  </body>
</html>
"""

HTML(string=html_string).write_pdf("output.pdf")
```
